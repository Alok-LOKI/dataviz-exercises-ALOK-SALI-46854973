# Lecture 02 – Data Visualization Exercise

## Objective
Analyze a dataset using Plotly to identify trends and patterns.

```
import pandas as pd
import plotly.express as px
```

```
df = px.data.gapminder()
df.head()
```

```
df.info()
df.describe()
```

## Data Overview
The dataset contains country, continent, population, GDP, and life expectancy.

```
india = df[df['country'] == 'India']
fig = px.line(india, x='year', y='lifeExp', title='Life Expectancy in India')
fig.show()
```

### Insight
Life expectancy increases over time.

```
df_2007 = df[df['year'] == 2007]
fig = px.bar(df_2007.sort_values(by='gdpPercap', ascending=False).head(10),
             x='country', y='gdpPercap',
             title='Top GDP Countries')
fig.show()
```

### Insight
Wealth is concentrated in few countries.

```
fig = px.pie(df_2007, names='continent', values='pop',
                  title='Population by Continent')
fig.show()
```

### Insight
Asia has the highest population.

```
fig = px.scatter(df_2007, x='gdpPercap', y='lifeExp',
                      size='pop', color='continent',
                      title='GDP vs Life Expectancy')
fig.show()
```

### Insight
Higher GDP relates to higher life expectancy.

## Conclusion
Visualization helps understand global trends.
