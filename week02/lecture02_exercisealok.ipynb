## Bar Charts: World Happiness Report 2023

```
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

np.random.seed(42)
regions=['Western Europe','Latin America','East Asia','Sub-Saharan Africa','South Asia']
rows=[]
for i in range(30):
    r=regions[i%5]
    base={'Western Europe':7.1,'Latin America':6.1,'East Asia':6.0,'Sub-Saharan Africa':4.5,'South Asia':4.8}[r]
    rows.append([f'Country {i+1}',r,round(base+np.random.normal(0,.3),2),round(np.random.uniform(.4,1),2),round(np.random.uniform(.4,1),2)])
df=pd.DataFrame(rows,columns=['Country','Region','Happiness_Score','GDP','Freedom'])
df.head()
```

```
region_avg=df.groupby('Region')['Happiness_Score'].mean().reset_index().sort_values('Happiness_Score')
fig=px.bar(region_avg,x='Happiness_Score',y='Region',orientation='h',text='Happiness_Score',
title='Western Europe leads average happiness while Sub-Saharan Africa trails')
fig.update_layout(template='plotly_white',xaxis=dict(range=[0,8]))
fig.show()
```

```
top8=df.nlargest(8,'Happiness_Score').copy()
top8['Group']='Top 8'
bottom8=df.nsmallest(8,'Happiness_Score').copy()
bottom8['Group']='Bottom 8'
combined=pd.concat([bottom8,top8]).sort_values('Happiness_Score')
fig=px.bar(combined,x='Happiness_Score',y='Country',color='Group',orientation='h',
title='The happiest countries score far above the least happy nations')
fig.update_layout(template='plotly_white',xaxis=dict(range=[0,8]))
fig.show()
```

```
sub=df.groupby('Region')[['GDP','Freedom']].mean().reset_index()
fig=go.Figure()
fig.add_bar(name='GDP',x=sub['Region'],y=sub['GDP'])
fig.add_bar(name='Freedom',x=sub['Region'],y=sub['Freedom'])
fig.update_layout(barmode='group',template='plotly_white',
title='Western Europe leads GDP while freedom varies across regions')
fig.show()
```
